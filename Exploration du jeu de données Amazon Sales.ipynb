# Import Required Libraries

Import necessary libraries such as pandas, numpy, matplotlib, and seaborn for data manipulation and visualization.

```python
import pandas as pd
```

```python
df=pd.read_csv("amazon.csv")
df.head()
```

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
# 1) Repérer les valeurs de 'rating' qui contiennent autre chose que chiffres/point/signe
mask_bad_rating = df["rating"].astype(str).str.contains(r"[^0-9\.\-]", na=False)

# 2) Voir les valeurs fautives et leur fréquence
df.loc[mask_bad_rating, "rating"].value_counts().head(20)
```

```python
import pandas as pd

num_cols = ["discounted_price", "actual_price", "discount_percentage", "rating", "rating_count"]

# 1) Garder seulement chiffres / signe / point, tout le reste → supprimé
df[num_cols] = (
    df[num_cols]
    .apply(lambda s: s.astype(str).str.replace(r"[^0-9.\-]", "", regex=True))
    .apply(pd.to_numeric, errors="coerce")  # 2) convertit en nombres, valeurs bizarres → NaN
)

# Vérification
print(df[num_cols].dtypes)
print("\nNaN par colonne :\n", df[num_cols].isna().sum())
```

```python
df[num_cols] = df[num_cols].fillna(df[num_cols].median())
```

```python
df[num_cols].isna().sum()
```

```python
# Statistiques de base
df[num_cols].describe()

# Prix moyen et taux de remise moyen
print("\nPrix moyen :", round(df["actual_price"].mean(), 2))
print("Prix moyen après remise :", round(df["discounted_price"].mean(), 2))
print("Remise moyenne (%) :", round(df["discount_percentage"].mean(), 2))
print("Note moyenne :", round(df["rating"].mean(), 2))

```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12,5))
sns.histplot(df["actual_price"], bins=30, kde=True)
plt.title("Distribution des prix avant remise")
plt.show()

plt.figure(figsize=(12,5))
sns.histplot(df["discount_percentage"], bins=30, kde=True, color='orange')
plt.title("Distribution des pourcentages de remise")
plt.show()

plt.figure(figsize=(12,5))
sns.histplot(df["rating"], bins=20, kde=True, color='green')
plt.title("Distribution des notes produits")
plt.show()
```

```python
# Regrouper par catégorie et calculer les moyennes
cat_summary = (
    df.groupby("category")[["actual_price", "discounted_price", "discount_percentage", "rating"]]
    .mean()
    .sort_values("actual_price", ascending=False)
)

cat_summary.round(2)
```

```python
cat_summary = (
    df.groupby("category")
    .agg(
        avg_price=("actual_price", "mean"),
        avg_discount=("discount_percentage", "mean"),
        avg_rating=("rating", "mean"),
        count=("product_id", "count")
    )
    .sort_values("avg_price", ascending=False)
)

cat_summary.round(2).head(10)
```

```python
plt.figure(figsize=(15,5))
for i, col in enumerate(["actual_price", "discount_percentage", "rating"], 1):
    plt.subplot(1, 3, i)
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot de {col}")
plt.tight_layout()
plt.show()
```

```python
Q1 = df["actual_price"].quantile(0.25)
Q3 = df["actual_price"].quantile(0.75)
IQR = Q3 - Q1

# Définir les limites
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Bornes : {lower_bound:.2f}  →  {upper_bound:.2f}")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Trouver la valeur limite (95e percentile)
price_cap = df["actual_price"].quantile(0.95)
print("Seuil visuel :", round(price_cap, 2))

# Visualiser jusqu'à ce seuil
plt.figure(figsize=(10,5))
sns.histplot(df[df["actual_price"] <= price_cap]["actual_price"], bins=30, kde=True)
plt.title("Distribution des prix (coupée à 95e percentile)")
plt.xlabel("Prix (₹)")
plt.show()
```

```python
# Top 10 catégories les plus chères
top10_price = (
    df.groupby("category")["actual_price"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,5))
sns.barplot(x=top10_price.values, y=top10_price.index, palette="Blues_r")
plt.title("Top 10 catégories par prix moyen")
plt.xlabel("Prix moyen (₹)")
plt.ylabel("Catégorie")
plt.show()
```
